# Basic Claude API call with Haiku

A minimal example: send one prompt to **Claude Haiku 4.5** and print the reply.

Haiku is the cheapest model ($1 / $5 per million input/output tokens) — ideal for learning and simple queries.

## 2. Set your API key

Best practice: set `ANTHROPIC_API_KEY` in your shell **before** launching Jupyter, so the key never lands in the notebook file.

```bash
export ANTHROPIC_API_KEY="sk-ant-..."
```

If you started Jupyter before setting it, the cell below is a fallback (paste-and-run, then clear the output). The `getpass` prompt keeps the key off-screen.

````python
import os

# Only prompt if the key isn't already in the environment.
if not os.environ.get("ANTHROPIC_API_KEY"):
    import getpass
    os.environ["ANTHROPIC_API_KEY"] = getpass.getpass("Paste your Anthropic API key: ")
````

## 3. Make the call

`Anthropic()` reads the key from the environment automatically.

````python
from anthropic import Anthropic

client = Anthropic()

response = client.messages.create(
    model="claude-haiku-4-5-20251001",
    max_tokens=300,
    messages=[
        {"role": "user", "content": "In two sentences, what is a large language model?"}
    ],
)

print(response.content[0].text)
````

## 4. See what it cost

Every response carries a `usage` object with the token counts. Handy for teaching students to watch spend.

````python
u = response.usage

# Haiku 4.5 rates, USD per million tokens
IN_RATE, OUT_RATE = 1.00, 5.00
cost = (u.input_tokens * IN_RATE + u.output_tokens * OUT_RATE) / 1_000_000

print(f"Input tokens:  {u.input_tokens}")
print(f"Output tokens: {u.output_tokens}")
print(f"Cost:          ${cost:.6f}")
````

## 5. Wrap it in a helper

A small reusable function so students can ask anything in one line. Renders the reply as Markdown.

````python
from IPython.display import Markdown

def ask(prompt, model="claude-haiku-4-5-20251001", max_tokens=500):
    r = client.messages.create(
        model=model,
        max_tokens=max_tokens,
        messages=[{"role": "user", "content": prompt}],
    )
    return Markdown(r.content[0].text)

ask("What is metformin?")
````
